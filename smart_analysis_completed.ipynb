```python
for i in range(len(usage_hours)):
    total = total + usage_hours[i]

mean = total / len(usage_hours)
print(f"For Mean: ${mean:.2f}")
```

```python
#While Loop
i=0
while i < len(usage_hours):
    total = total + usage_hours[i]
    i = i + 1
    print(i) # i will not show if not reset to 0 in the beginning
    
mean = total / len(usage_hours)
mean = round(mean,1)
print("While Mean :", mean)
```

### Use of Smartphone relates to education performance?

```python
df2 = pd.read_csv('../data/Impact_of_Mobile_Phone_on_Students_Health_Cleaned.csv')
summary=[]

for usage_level in df2['Mobile phone use for education'].unique():
    subset = df2[df2['Mobile phone use for education'] == usage_level]
    yes_count = len(subset[subset['Helpful for studying'] == 'Yes'])
    total = len(subset)
    percent_helpful = round((yes_count / total) * 100, 1) if total > 0 else 0

    if percent_helpful >= 70:
        verdict = 'Strongly helpful'
    elif percent_helpful >= 40:
        verdict = 'Somewhat helpful'
    else:
        verdict = 'Not very helpful'

    summary[usage_level] = {
        'students': total,
        'percent_found_helpful': percent_helpful,
        'verdict': verdict
    }

for level, result in summary.items():
    print(level, '->', result)
```
